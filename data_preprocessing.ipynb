Removing Columns where null values are more than 50%

```python
null_percentage = (data.isnull().mean() * 100).round(2)
print('Null value Percentage Column Wise: ',null_percentage)
total_cols = len(data.columns)
```

Removing Leakage columns that are not either identifiers or not needed for any further tasks

```python
targets = ["EnergyRating", "BerRating", "CO2Rating", "MainSpaceHeatingFuel"]

calc_outputs = ["RER", "RenewEPnren", "RenewEPren", "CPC", "EPC", "MPCDERValue",
    "SHMainPrimaryEnergy", "SHSecPrimaryEnergy", "WHMainPrimaryEnergy", "WHSecPrimaryEnergy",
    "PumpsFansPrimaryEnergy", "LightingPrimaryEnergy", "CoolingPrimaryEnergy",
    "CHPInputPrimaryEnergy", "CHPElecOutputPrimaryEnergy",
    "RenewPrimaryEnergy1", "RenewPrimaryEnergy2", "RenewPrimaryEnergy3", "RenewPrimaryEPTot"]   

energy_blocks = [c for c in data.columns if c.startswith(
    ("FirstEner", "SecondEner", "ThirdEner", "FirstPartL", "SecondPartL", "ThirdPartL"))]      
```

```python
to_drop = list(set(calc_outputs + energy_blocks + ["FirstWallTypeId"]))
assert not set(to_drop) & set(targets)

log = pd.DataFrame({"column": to_drop})          # save with reasons for the report
data = data.drop(columns=to_drop)
```

Fixing county names so all data of a county has a single regerence

```python
data["CountyName_raw"] = data["CountyName"]

data["CountyName"] = (
    data["CountyName"].str.strip()
    .str.replace(r"^Dublin\s*\d+W?$", "Co. Dublin", regex=True)
    .str.replace(r"^(Cork|Limerick|Galway|Waterford) City$", r"Co. \1", regex=True)
)

print(data["CountyName"].nunique())
print(data["CountyName"].value_counts())
```

Changing datatypes where nedded


YES/NO columns

```python
yn_cols = [c for c in data.select_dtypes("string").columns
           if set(data[c].dropna().str.strip().str.upper().unique()) <= {"YES", "NO"}]
for c in yn_cols:
    data[c] = data[c].str.strip().str.upper().map({"YES": 1, "NO": 0}).astype("Int8")
data["IsSolarPVBatteryPresent"] = data["IsSolarPVBatteryPresent"].astype("int8")
print("binary:", yn_cols)
```

Numeric codes that are categories

```python
code_cols = ["SupplSHFuel", "SupplWHFuel", "SHRenewableResources", "WHRenewableResources",
             "HeatSystemControlCat", "HeatSystemResponseCat", "TGDLEdition", "FirstWallAgeBandId"]
for c in code_cols:
    data[c] = data[c].astype("Int64").astype("category")
```

ordered categoricals

```python
ber_order = ["A1","A2","A3","B1","B2","B3","C1","C2","C3","D1","D2","E1","E2","F","G"]
mass_order = ["Low", "Medium-low", "Medium", "Medium-high", "High"]
data["EnergyRating"] = pd.Categorical(data["EnergyRating"], categories=ber_order, ordered=True)
data["ThermalMassCategory"] = pd.Categorical(data["ThermalMassCategory"], categories=mass_order, ordered=True)
print(data[["EnergyRating", "ThermalMassCategory"]].isna().sum())   # must be 0 new NaNs
```

```python
str_cols = [c for c in data.select_dtypes("string").columns if c != "FirstWallDescription"]
data[str_cols] = data[str_cols].astype("category")
```

```python
for c in data.select_dtypes("float64").columns:
    data[c] = pd.to_numeric(data[c], downcast="float")
for c in data.select_dtypes("int64").columns:
    data[c] = pd.to_numeric(data[c], downcast="integer")
```

Noe we check if the data in each attribute correspond to the charactoeristics

```python
TOT = "GroundFloorArea(sq m)"         
has = lambda cols: [c for c in cols if c in data.columns]
t1 = pd.DataFrame(index=data.index)   
t2 = pd.DataFrame(index=data.index)   
add = lambda df, name, m: df.__setitem__(name, m.fillna(False).astype(bool))
```

```python
add(t1, "year<1700|>2026", (data["Year_of_Construction"] < 1700) | (data["Year_of_Construction"] > 2026))
add(t1, "built>2yrs after assessment", data["Year_of_Construction"] > data["DateOfAssessment"].dt.year + 2)

non_neg = has([TOT, "WallArea", "RoofArea", "FloorArea", "WindowArea", "DoorArea",
    "GroundFloorArea", "FirstFloorArea", "SecondFloorArea", "ThirdFloorArea", "RoomInRoofArea",
    "PredominantRoofTypeArea", "FirstWallArea", "GroundFloorHeight", "FirstFloorHeight",
    "SecondFloorHeight", "ThirdFloorHeight", "PermeabilityTestResult", "WaterStorageVolume",
    "InsulationThickness", "ThermalBridgingFactor", "HeatLossIndicator", "NoOfChimneys",
    "NoOfOpenFlues", "NoOfFansAndVents", "NoOfFluelessGasFires", "NoOfSidesSheltered",
    "NoCentralHeatingPumps", "NoOilBoilerHeatingPumps", "NoGasBoilerHeatingPumps",
    "HSMainSystemEfficiency", "WHMainSystemEff", "HSSupplSystemEff"])
for c in non_neg:
    add(t1, f"{c}<0", data[c] < 0)

for c in has(["UValueWall", "UValueRoof", "UValueFloor", "UValueWindow", "UvalueDoor",
              "GroundFloorUValue", "FirstWallUValue"]):
    add(t1, f"{c} <0|>10", (data[c] < 0) | (data[c] > 10))

for c in has(["GroundFloorHeight", "FirstFloorHeight", "SecondFloorHeight", "ThirdFloorHeight"]):
    add(t1, f"{c}>10", data[c] > 10)
for c in has(["HSMainSystemEfficiency", "WHMainSystemEff", "HSSupplSystemEff"]):
    add(t1, f"{c}>1000", data[c] > 1000)          # heat pumps can exceed 100%
for c in has(["PercentageDraughtStripped", "LivingAreaPercent"]):
    add(t1, f"{c} outside 0-100", (data[c] < 0) | (data[c] > 100))
if "HSSupplHeatFraction" in data: add(t1, "HSSupplHeatFraction outside 0-1", (data["HSSupplHeatFraction"] < 0) | (data["HSSupplHeatFraction"] > 1))

add(t1, f"{TOT}<=0", data[TOT] <= 0)
add(t1, "NoStoreys <1|>10", (data["NoStoreys"] < 1) | (data["NoStoreys"] > 10))
add(t1, "BerRating missing", data["BerRating"].isna())
add(t1, "CO2Rating missing", data["CO2Rating"].isna())
add(t1, "EnergyRating missing", data["EnergyRating"].isna())
```

```python
bins = [-np.inf, 25, 50, 75, 100, 125, 150, 175, 200, 225, 260, 300, 340, 380, 450, np.inf]
labels = ["A1","A2","A3","B1","B2","B3","C1","C2","C3","D1","D2","E1","E2","F","G"]
expected = pd.cut(data["BerRating"], bins=bins, labels=labels, right=True)
add(t1, "EnergyRating != band of BerRating", expected.astype("object") != data["EnergyRating"].astype("object"))
```

```python
add(t2, f"{TOT} <10|>1000", (data[TOT] < 10) | (data[TOT] > 1000))
add(t2, "BerRating <-200|>1500", (data["BerRating"] < -200) | (data["BerRating"] > 1500))
add(t2, "CO2Rating <-200|>1000", (data["CO2Rating"] < -200) | (data["CO2Rating"] > 1000))
add(t2, "WindowArea > WallArea", data["WindowArea"] > data["WallArea"])
fl = data[has(["GroundFloorArea", "FirstFloorArea", "SecondFloorArea", "ThirdFloorArea"])].fillna(0).sum(axis=1)
add(t2, "sum of floor areas != total (>1 m2)", (fl - data[TOT]).abs() > 1)
```

```python
def report(flags, title):
    cnt = flags.sum().sort_values(ascending=False)
    print(f"\n{title}\n{cnt[cnt > 0].to_string()}")
    n = flags.any(axis=1).sum()
    print(f"\nRows with at least one flag: {n:,} of {len(data):,} ({n/len(data):.2%})")

report(t1, "TIER 1 (impossible)")
report(t2, "TIER 2 (suspicious, keep)")
```

```python
cols_to_keep = [
    # Targets
    "EnergyRating", "BerRating", "CO2Rating", "MainSpaceHeatingFuel",

    # Dwelling descriptors
    "CountyName", "DwellingTypeDescr", "Year_of_Construction", "NoStoreys",

    # Fabric and size
    "GroundFloorArea(sq m)", "UValueWall", "UValueRoof", "UValueFloor",
    "UValueWindow", "UvalueDoor", "WallArea", "RoofArea", "FloorArea",
    "WindowArea", "DoorArea", "StructureType", "PredominantRoofType",
    "RoomInRoofArea", "LivingAreaPercent",

    # Heating, hot water and ventilation
    "MainWaterHeatingFuel", "HSMainSystemEfficiency", "WHMainSystemEff",
    "HSSupplHeatFraction", "HeatSystemControlCat", "HeatSystemResponseCat",
    "NoOfChimneys", "NoOfOpenFlues", "NoOfFansAndVents", "VentilationMethod",
    "DraftLobby", "PercentageDraughtStripped", "NoOfSidesSheltered",
    "SuspendedWoodenFloor", "InsulationType", "InsulationThickness",
    "WaterStorageVolume", "SolarHotWaterHeating", "SolarPVEnergyOutput",
]

data = data[cols_to_keep].copy()

print(data.shape)  
```
